装饰器

被修饰函数：
    无参数
    无返回值
装饰器：
    无参数

```python
def decorateFunc(func):
    def wrapper():
        print("before func")
        func()
        print("after func")
    return wrapper

@decorateFunc
def test1():
    print("test1")
    
test1()
```

被修饰函数：
    有参数
    无返回值
装饰器：
    无参数

```python
def decorateFunc(func):
    def wrapper(*args,**kwargs):
        print("before func")
        func(*args,**kwargs)
        print("after func")
    return wrapper

@decorateFunc
def test2(a,b):
    print(a+b)
    
test2(1,2)
```

```python
def decorateFunc(func):
    def wrapper(*args,**kwargs):
        print("before func")
        res = func(*args,**kwargs)
        print("after func")
        return res
    return wrapper

@decorateFunc
def test3(a,b):
    return a+b
    
print(test3(1,2))
```

装饰器：
被修饰函数：
    有参数
    有返回值
装饰器：
    有参数

```python
def decorateFunc(times):
    def nextLevel(func):
        def wrapper(*args,**kwargs):
            for i in range(times):
                print(i)
            print("before func")
            res = func(*args,**kwargs)
            print("after func")
            return res
        return wrapper
    return nextLevel

@decorateFunc(3)
def test3(a,b):
    return a+b
    
print(test3(1,2))
```
